## Declaración de variables

Consideramos el siguiente modelo, que combina una variable continua no negativa $x$, una entera no negativa $n$, una binaria $y$ y una continua libre $z$:

$$
\begin{aligned}
\max\quad & 3x+2n-y\\
\text{s.a}\quad & x+2n\leq 7,\\
& x\leq 2.5y,\\
& z=n-x,\\
& x\geq 0,\quad n\in\mathbb{Z}_{\geq 0},\quad
  y\in\{0,1\},\quad z\in\mathbb{R}.
\end{aligned}
$$

```python
from docplex.mp.model import Model

# 1. Crear el modelo
mdl = Model(name="tipos_de_variables")

# 2. Declarar las variables
x = mdl.continuous_var(lb=0, name="x")
n = mdl.integer_var(lb=0, name="n")
y = mdl.binary_var(name="y")
z = mdl.continuous_var(lb=-mdl.infinity, name="z")

# 3. Escribir la función objetivo
mdl.maximize(3 * x + 2 * n - y)

# 4. Añadir las restricciones
mdl.add_constraint(x + 2 * n <= 7)
mdl.add_constraint(x <= 2.5 * y)
mdl.add_constraint(z == n - x)

# 5. Resolver y mostrar la solución
sol = mdl.solve()
print(sol)
```

## Resolver la relajación lineal

La relajación lineal permite que las variables enteras y binarias tomen valores continuos, conservando sus cotas y todas las restricciones. Las binarias pasan a ser continuas entre $0$ y $1$.

A partir del modelo `tipos_de_variables`, obtenemos su relajación mediante `LinearRelaxer`, la resolvemos y mostramos la solución. Ejecuta primero la celda del ejemplo de declaración de variables para que `mdl` contenga ese modelo.

```python
from docplex.mp.relax_linear import LinearRelaxer

mdl_rl = LinearRelaxer.make_relaxed_model(
    mdl, relaxed_name=f"{mdl.name}_rl"
)
sol_rl = mdl_rl.solve()
print(sol_rl)
```
