Consideremos el siguiente modelo con variables continuas:

$$
\begin{aligned}
\text{máx}\quad & 3x+2y\\
\text{s.a}\quad
& x+y\leq 4,\\
& 0\leq x\leq 2,\\
& 0\leq y\leq 3.
\end{aligned}
$$

```python
from docplex.mp.model import Model

# 1. Crear el modelo
mdl = Model(name="primer_modelo")

# 2. Declarar las variables y sus cotas
x = mdl.continuous_var(lb=0, ub=2, name="x")
y = mdl.continuous_var(lb=0, ub=3, name="y")

# 3. Escribir la función objetivo
mdl.maximize(3 * x + 2 * y)

# 4. Añadir las restricciones
mdl.add_constraint(x + y <= 4, ctname="recurso")

# 5. Resolver y mostrar la solución
sol = mdl.solve()
print(sol)
```
